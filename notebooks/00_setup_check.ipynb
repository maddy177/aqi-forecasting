```python
import numpy as np, pandas as pd, matplotlib, seaborn, sklearn, scipy, statsmodels, torch
import torch.nn as nn
from statsmodels.tsa.arima.model import ARIMA

print("pandas", pd.__version__, "| statsmodels", statsmodels.__version__, "| torch", torch.__version__)

# a tiny ARIMA on a made-up series
rng = np.random.default_rng(0)
y = pd.Series(np.cumsum(rng.normal(size=200)))
res = ARIMA(y, order=(1, 1, 1)).fit()
print("ARIMA forecast:", res.forecast(3).round(2).tolist())

# a tiny LSTM, one forward pass
lstm = nn.LSTM(input_size=1, hidden_size=8, batch_first=True)
out, _ = lstm(torch.randn(4, 10, 1))
print("LSTM output shape:", tuple(out.shape))   # expect (4, 10, 8)
```
